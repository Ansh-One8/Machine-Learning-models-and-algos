# Stopwords 
are common words in a language (like "the", "is", "in", "and", "a", "of") that occur very frequently in text but carry little to no meaningful/distinctive information for many NLP tasks. They're typically removed during text preprocessing to reduce noise and focus on the words that actually carry meaning.
example:-

Original: "The cat is sitting on the mat"
After stopword removal: "cat sitting mat"


#### Purpose:
Reduce dimensionality/noise in text data before feeding it into models (classification, search, topic modeling, etc.)
High frequency, low information: Words like "the", "is", "at" appear everywhere but don't help distinguish one document/sentence from another
Language-specific: Every language has its own stopword list (English stopwords ≠ Hindi stopwords)
Not always removed: For tasks like sentiment analysis, machine translation, or POS tagging, stopwords can actually matter (e.g., "not good" vs "good" — removing "not" changes meaning!)
Customizable: You can add/remove words from the default stopword list based on your domain (e.g., "app", "click" might be stopwords in a mobile-app-reviews dataset)

```python
import nltk
nltk.download('wordnet')
nltk.download('omw-1.4')
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
```

```python
# applying the Stopwords first then filter and then stemming
## Apply Stopwords And Filter And then Apply Snowball Stemming

for i in range(len(sentences)):
    sentences[i] = sentences[i].lower()
    words = nltk.word_tokenize(sentences[i])
    words = [lemmatizer.lemmatize(word, pos='v') for word in words if word not in set(stopwords.words('english'))]
    sentences[i] = ' '.join(words)  # converting all the list of words into sentences
    
```
