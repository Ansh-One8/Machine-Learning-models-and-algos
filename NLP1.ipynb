# Defination

- **Corpus** – A collection of multiple text documents.
- **Document** – A single unit of text (e.g., one review, email, or SMS) within a corpus.
- **Paragraph** – A group of related sentences within a document.
- **Sentence** – A grammatically complete unit of text ending in `.`, `!`, or `?`.
- **Word/Token** – The smallest meaningful unit of text (usually a word) obtained after tokenization.
- **Character** – A single letter, digit, or symbol — the most granular unit of text.
- **Tokenization** – The process of splitting text into smaller units (sentences, words, or characters).
- **Stopwords** – Common, low-information words (e.g., "is", "the", "and") usually removed during preprocessing.
- **Stemming** – Chopping words down to their root form using rules, which may not be a real word (e.g., "studies" → "studi").
- **Lemmatization** – Reducing words to their proper dictionary base form using vocabulary and grammar (e.g., "studies" → "study").
- **POS Tagging** – Labeling each word with its grammatical role (noun, verb, adjective, etc.).
- **Named Entity Recognition (NER)** – Identifying real-world entities (person, place, organization, date) in text.
- **One Hot Encoding (OHE)** – Representing each word as a binary vector with a single "1" marking its position.
- **Bag of Words (BoW)** – Representing text as word frequency counts, ignoring grammar and word order.
- **Vocabulary** – The set of all unique words present in a corpus.
- **N-gram** – A contiguous sequence of N words from text (e.g., bigram = 2 words, trigram = 3 words).

```python
import nltk
nltk.download('wordnet')
nltk.download('omw-1.4')
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
lemmatizer.lemmatize("going", pos="v")
```
